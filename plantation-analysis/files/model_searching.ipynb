```python
import pandas as pd

df = pd.read_excel("Selected Sheets 2.xlsx")

df.info()
```

```python
df.head()
```

```python
df = df.drop(columns = ["No", "Keterangan"])

df["Nomor Baris"].value_counts()
```

```python
df["Nomor Baris"] = df["Nomor Baris"].fillna(25.0)

df["Nomor Tanam"].value_counts()
```

```python
df["Nomor Tanam"] = df["Nomor Tanam"].fillna(1.0)

df["Jumlah Sisir"].value_counts()
```

```python
df["Jumlah Sisir"] = df["Jumlah Sisir"].fillna(9.0)

df["Kaliper"].value_counts()
```

```python
df["Kaliper"] = df["Kaliper"].fillna(38.0)

import matplotlib.pyplot as plt

df.boxplot(column='Bruto/Sisir')
plt.show()
```

```python
import numpy as np

df["Bruto/Sisir"] = df["Bruto/Sisir"].fillna(np.nanmedian(df["Bruto/Sisir"]))

df.info()
```

```python
df['Bruto (Kg)'] = pd.to_numeric(df["Bruto (Kg)"], errors = "coerce")

df["Clone"].value_counts()
```

```python
df = df.drop(columns=["Clone"])

df["Generasi"].value_counts()
```

```python
df["Generasi"] = df["Generasi"].str.lower()

df["Generasi"].value_counts()
```

```python
df["Generasi"] = df["Generasi"].str.strip()

df["Generasi"].value_counts()
```

```python
mapping = {
    "induk": 0,
    "r1": 1,
    "r2": 2,
    "r3": 3,
    "r4": 4
}
df['Generasi'] = df['Generasi'].map(mapping)
df['Tanggal Panen'] = pd.to_datetime(df['Tanggal Panen'])
df['Tahun Panen'] = df['Tanggal Panen'].dt.year
df['Bulan Panen'] = df['Tanggal Panen'].dt.month
df['Hari Panen'] = df['Tanggal Panen'].dt.day
df = df.drop(columns=["Tanggal Panen"])

df.info()
```

```python
df.boxplot(column='Bruto (Kg)')
plt.show()
```

```python
df["Bruto (Kg)"] = df["Bruto (Kg)"].fillna(np.nanmedian(df["Bruto (Kg)"]))

df.info()
```

```python
from sklearn.model_selection import train_test_split

x = df.drop(columns = ["Bruto/Sisir"])
y = df["Bruto/Sisir"]
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import RadiusNeighborsRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    root_mean_squared_error,
    r2_score
)
import numpy as np

# Fit the scaler only on training data
scaler = StandardScaler()

x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau

tf.keras.utils.set_random_seed(42)

model = Sequential([
    Input(shape = (x_train_scaled.shape[1],)),
    Dense(128, activation = "relu"),
    Dropout(0.2),
    Dense(64, activation = "relu"),
    Dropout(0.1),
    Dense(32, activation = "relu"),
    Dense(1, activation = "linear")
])

model.summary()
```

```python
model.compile(
    optimizer = tf.keras.optimizers.Adam(learning_rate = 0.001),
    loss = "mse",
    metrics = [
        tf.keras.metrics.MeanAbsoluteError(name = "mae"),
        tf.keras.metrics.MeanAbsolutePercentageError(name = "mape"),
        tf.keras.metrics.MeanSquaredError(name = "mse"),
        tf.keras.metrics.RootMeanSquaredError(name = "rmse"),
        tf.keras.metrics.R2Score(name = "r2")
    ]
)
```

```python
early_stopping = EarlyStopping(
    monitor = "val_loss",
    patience = 20,
    restore_best_weights = True,
    verbose = 1
)
reduce_lr = ReduceLROnPlateau(
    monitor = "val_loss",
    factor = 0.5,
    patience = 8,
    min_lr = 1e-6,
    verbose = 1
)
```

```python
history = model.fit(
    x_train_scaled,
    y_train,
    validation_split = 0.2,
    epochs = 300,
    batch_size = 32,
    callbacks = [
        early_stopping,
        reduce_lr
    ],
    verbose = 1
)
```

```python
y_pred = model.predict(x_test_scaled).ravel()

print("Metrics:")
print(f"MAE: {mean_absolute_error(y_test, y_pred)}")
print(f"MSE: {mean_squared_error(y_test, y_pred)}")
print(f"RMSE: {root_mean_squared_error(y_test, y_pred)}")
print(f"R2: {r2_score(y_test, y_pred)}")
```

```python
test_results = model.evaluate(
    x_test_scaled,
    y_test,
    verbose = 0
)

print(dict(zip(model.metrics_names, test_results)))
```

```python
import keras_tuner as kt
from tensorflow import keras
from tensorboard.plugins.hparams import api as hp

def build_model(hp):
    model = keras.Sequential()

    model.add(
        keras.layers.Input(shape=(x_train_scaled.shape[1],))
    )

    num_layers = hp.Int(
        "num_layers",
        min_value=1,
        max_value=3
    )

    for i in range(num_layers):
        model.add(
            keras.layers.Dense(
                units=hp.Choice(
                    f"units_{i}",
                    [32, 64, 128]
                ),
                activation=hp.Choice(
                    f"activation_{i}",
                    ["relu", "tanh"]
                )
            )
        )
        model.add(
            keras.layers.Dropout(
                hp.Choice(
                    f"dropout_{i}",
                    [0.0, 0.1, 0.2, 0.3]
                )
            )
        )

    model.add(
        keras.layers.Dense(1)
    )

    optimizer_name = hp.Choice(
        "optimizer",
        ["adam", "rmsprop", "sgd"]
    )

    learning_rate = hp.Choice(
        "learning_rate",
        [1e-2, 1e-3, 5e-4, 1e-4]
    )

    if optimizer_name == "adam":
        optimizer = keras.optimizers.Adam(
            learning_rate=learning_rate
        )
    elif optimizer_name == "rmsprop":
        optimizer = keras.optimizers.RMSprop(
            learning_rate=learning_rate
        )
    else:
        optimizer = keras.optimizers.SGD(
            learning_rate=learning_rate
        )

    model.compile(
        optimizer=optimizer,
        loss="mse",
        metrics=[
            keras.metrics.MeanAbsoluteError(name="mae"),
            keras.metrics.RootMeanSquaredError(name="rmse")
        ]
    )

    return model
```

```python
tuner = kt.GridSearch(
    hypermodel = build_model,
    objective = "val_loss",
    max_trials = 100,
    directory = "tuning",
    project_name = "banana_fcnn",
    overwrite = True
)
```

```python
tuner.search(
    x_train_scaled,
    y_train,
    validation_split = 0.2,
    epochs = 100,
    callbacks = [early_stopping, reduce_lr],
    verbose = 1
)
```

```python
best_model = tuner.get_best_models(1)[0]
best_hp = tuner.get_best_hyperparameters(1)[0]

print(best_hp.values)
```
