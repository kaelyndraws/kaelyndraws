```python
import pandas as pd

df = pd.read_csv("data_B.csv")

df.head()
```

Variabel target dalam proyek ini adalah "Credit_Score", sedangkan variabel lain merupakan fitur.

- "Month" mewakili bulan pelanggan mengumpulkan aplikasi peminjaman. "Month" dapat mempengaruhi "Credit_Score" melalui pembaharuan rasio pengunaan kredit oleh saldo kartu kredit, laporan pembayaran tepat waktu, dan penambahan usia kredit.
- "Age" mewakili usia pelanggan dalam ukuran tahun. "Age" dapat mempengaruhi "Credit_Score" karena pelanggan yang lebih tua cenderung memiliki skor kredit yang lebih tinggi.
- "Occupation" mewakili pekerjaan yang dipegang oleh pelanggan. Secara tidak langsung, "Occupation" dapat mempengaruhi "Credit_Score" melalui kestabilan pemasukan, pemeriksaan kredit untuk pekerjaan, asuransi industri, dan suku bunga.
- "Annual_Income" mewakili pemasukan tahunan pelanggan. Secara tidak langsung, "Annual_Income" mempengaruhi "Credit_Score" dengan menghindar jejak negatif, perbandingan antar hutang dan pemasukan, dan limit kredit.
- "Monthly_Inhand_Salary" mewakili gaji bulanan pelanggan. Secara tidak langsung, "Monthly_Inhand_Salary" mempengaruhi "Credit_Score" melalui perbandingan antar hutang dan pemasukan, penerimaan limit kredit, dan pembayaran tepat waktu.
- "Num_Bank_Accounts" mewakili jumlah rekening bank milik pelanggan. Secara tidak langsung, "Num_Bank_Accounts" mempengaruhi "Credit_Score" melalui aplikasi kredit, penggunaan *overdraft*, jumlah rekening kartu kredit, dan kebiasaan finansial yang positif.
- "Num_Credit_Card" mewakili jumlah kartu kredit milik pelanggan. "Num_Credit_Card" dapat mempengaruhi "Credit_Score" karena berkontribusi pada total kredit yang tersedia, rasio penggunaan kredit, jumlah *hard inquiry*, dan usia rekening kredit.
- "Interest_Rate" mewakili suku bunga pelanggan. Secara tidak langsung, "Interest_Rate" mempengaruhi "Credit_Score" melalui pembayaran minimum dan penggunaan kredit.
- "Num_of_Loan" mewakili jumlah hutang pelanggan. "Num_of_Loan" dapat mempengaruhi "Credit_Score" karena membentuk sejarah pembayaran, *credit mix*, dan perbandingan antar hutang dan pemasukan.
- "Type_of_Loan" mewakili jenis hutang pelanggan. "Type_of_Loan" dapat mempengaruhi "Credit_Score" melalui *credit mix*, rasio penggunaan kredit, dan *hard inquiry*.
- "Delay_from_due_date" mewakili panjang waktu pelanggan menunda pembayaran.
- "Num_of_Delayed_Payment" mewakili jumlah pembayaran tertunda pelanggan.
- "Changed_Credit_Limit" mewakili limit kredit baru yang diajukan pelanggan. "Changed_Credit_Limit" dapat mempengaruhi "Credit_Score" dengan mengubah rasio penggunaan kredit.
- "Num_Credit_Inquiries" mewakili jumlah pemeriksaan kredit pelanggan.
- "Credit_Mix" mewakili keragaman akun kredit pelanggan.
- "Outstanding_Debt" mewakili hutang pelanggan. "Outstanding_Debt" dapat mempengaruhi "Credit_Score" melalui pemeriksaan rasio penggunaan kredit.
- "Credit_Utilization_Ratio" mewakili jumlah limit kredit yang digunakan pelanggan.
- "Credit_History_Age" mewakili usia rata-rata akun kredit pelanggan. "Credit_History_Age" dapat mempengaruhi "Credit_Score" dengan menunjukkan kestabilan finansial pelanggan.
- "Payment_of_Min_Amount" mewakili kenyataan bahwa pelanggan membayar kembali hutang kredit. "Payment_of_Min_Amount" dapat mempengaruhi "Credit_Score" melalui sejarah pembayaran dan penggunaan kredit.
- "Total_EMI_per_month" mewakili angsuran belanan pelanggan. "Total_EMI_per_month" dapat mempengaruhi "Credit_Score" melalui penentuan *Fixed Obligation to Income Ratio* (FOIR) dan sejarah pembayaran.
- "Amount_invested_monthly" mewakili jumlah investasi bulanan pelanggan. Secara tidak langsung, "Amount_invested_monthly" mempengaruhi "Credit_Score" melalui sejarah pembayaran, penggunaan kredit, dan kemungkinan pelanggan meminjam uang demi investasi.
- "Payment_Behavior" mewakili sifat pembayaran pelanggan.
- "Monthly_Balance" mewakili saldo bulanan pelanggan. "Monthly_Balance" dapat mempengaruhi "Credit_Score" melalui penggunaan kredit.

##### One-Hot Encode

```python
x_train_og = x_train.copy()
x_train_og["Payment_Behaviour"] = (
    x_train_og["Payment_Behaviour"]
    .replace("!@9#%8", "Unknown")
)
x_train_og = pd.get_dummies(
    x_train_og,
    columns = ["Payment_Behaviour"],
    drop_first = True,
    dtype = int
)

x_train_og.info()
```

##### Ordinal Encode

```python
spending_map = {
    "Low_spent_Small_value_payments": "Low",
    "Low_spent_Medium_value_payments": "Low",
    "Low_spent_Large_value_payments": "Low",
    "High_spent_Small_value_payments": "High",
    "High_spent_Medium_value_payments": "High",
    "High_spent_Large_value_payments": "High",
}
x_train_split = x_train.copy()
x_train_split["Spending_Level"] = x_train_split["Payment_Behaviour"].map(spending_map)
payment_map = {
    "Low_spent_Small_value_payments": "Small",
    "Low_spent_Medium_value_payments": "Medium",
    "Low_spent_Large_value_payments": "Large",
    "High_spent_Small_value_payments": "Small",
    "High_spent_Medium_value_payments": "Medium",
    "High_spent_Large_value_payments": "Large",
}
x_train_split["Payment_Size"] = x_train_split["Payment_Behaviour"].map(payment_map)
x_train_split["Spending_Level"] = x_train_split["Spending_Level"].fillna("Unknown")
x_train_split["Payment_Size"] = x_train_split["Payment_Size"].fillna("Unknown")
x_train_split["Spending_Level"] = x_train_split["Spending_Level"].map({
    "Low": 0,
    "High": 1,
    "Unknown": -1
})
x_train_split["Payment_Size"] = x_train_split["Payment_Size"].map({
    "Small": 0,
    "Medium": 1,
    "Large": 2,
    "Unknown": -1
})
x_train_split = x_train_split.drop(columns = ["Payment_Behaviour"])

x_train_split.info()
```

# Model Machine Learning & Eksperimen Mendapatkan Model Terbaik

## Pemilihan Model Machine Learning dan Mengapa Masing-Masing Model Dipilih

### *Random Forest*

*Random Forest* menggabungkan banyak *Decision Tree* dan *aggregate* prediksi mereka.

#### Kelebihan

- Mengurangi *overfitting*
- Menangani hubungan yang kompleks
- Menyediakan kepentingan fitur

#### Kekurangan

- Lebih sulit untuk diinterpretasikan daripada satu *tree*

### XGBoost

XGBoost membangun *tree* secara berurutan, dimana setiap *tree* baru memperbaiki kesalahan yang dibuat oleh *tree* sebelumnya.

#### Kelebihan

- Kinerja prediksi yang tinggi
- Menangani interaksi yang tidak linear
- Seringkali canggih untuk data tabel

#### Kekurangan

- Memakan banyak daya komputasi

## Persiapan Eksperimen

### Metrik Evaluasi

Kinerja model akan dievaluasi menggunakan akurasi, presisi, *recall*, dan *F1-Score* untuk memberikan penilaian komprehensif pada kinerja klasifikasi.

## Eksperimen Model

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix

# Before fitting, check for infinite or very large values
print(x_train_og.max().max())
print(x_train_split.max().max())

# Replace infinite values with NaN and then fill NaN with a reasonable value (e.g., median or a large but finite number)
x_train_og = x_train_og.replace([float('inf'), -float('inf')], float('nan'))
x_train_og = x_train_og.fillna(x_train_og.median())
x_train_split = x_train_split.replace([float('inf'), -float('inf')], float('nan'))
x_train_split = x_train_split.fillna(x_train_split.median())

# Also check test sets
x_test_og = x_test_og.replace([float('inf'), -float('inf')], float('nan'))
x_test_og = x_test_og.fillna(x_test_og.median())
x_test_split = x_test_split.replace([float('inf'), -float('inf')], float('nan'))
x_test_split = x_test_split.fillna(x_test_split.median())

# Align columns of test sets with training sets
x_test_og = x_test_og.reindex(columns=x_train_og.columns, fill_value=0)
x_test_split = x_test_split.reindex(columns=x_train_split.columns, fill_value=0)

rf_og = RandomForestClassifier(random_state = 42)
rf_split = RandomForestClassifier(random_state = 42)

rf_og.fit(x_train_og, y_train)
rf_split.fit(x_train_split, y_train)

y_pred_rf_og = rf_og.predict(x_test_og)
y_pred_rf_split = rf_split.predict(x_test_split)

print("Original Data")
print(classification_report(y_test, y_pred_rf_og))
print(confusion_matrix(y_test, y_pred_rf_og))
print("\n")
print("Split Data")
print(classification_report(y_test, y_pred_rf_split))
print(confusion_matrix(y_test, y_pred_rf_split))
```
